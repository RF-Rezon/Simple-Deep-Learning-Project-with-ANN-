```python
import numpy as np
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler, LabelEncoder
from sklearn.preprocessing import OneHotEncoder
import pickle
import matplotlib.pyplot as plt
```

```python
data = pd.read_csv('/Churn_Modelling.csv')
data
```

```python
data.isna().sum()
```

```python
data=data.drop(['RowNumber','CustomerId','Surname'], axis=1)
data
```

```python
label_encoder_gender = LabelEncoder()
data['Gender'] = label_encoder_gender.fit_transform(data['Gender'])
data
```

```python
data['Geography'].value_counts()
```

```python
onehot_encoder = OneHotEncoder()
geo_encoded  = onehot_encoder.fit_transform(data[['Geography']])
geo_encoded
```

```python
onehot_encoder.get_feature_names_out(['Geography'])
```

```python
geo_encoded_df = pd.DataFrame(geo_encoded.toarray(), columns=onehot_encoder.get_feature_names_out(['Geography']))
```

```python
data = pd.concat([data.drop('Geography', axis=1), geo_encoded_df], axis=1)
data
```

```python
with open(file='label_encoder_gender.pkl' , mode='wb') as file:
    pickle.dump(label_encoder_gender,file)


with open(file='one_hot_geo_encoded.pkl' , mode='wb') as file:
    pickle.dump(geo_encoded,file)
```

```python
X = data.drop('Exited', axis=1)
y = data['Exited']
```

```python
from sklearn.model_selection import train_test_split
```
